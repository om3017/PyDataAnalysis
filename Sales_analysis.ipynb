# **Data Cleaning**

- **Renaming Occupation as Sector**

```python
data.rename(columns={"Occupation":"Sector"},inplace=True)
```

```python
data
```

- **Replacing values- 0 and 1 to single and married respectively in Marital Status column**

```python
data["Marital_Status"] = data["Marital_Status"].astype(str).str.replace("1", "Married")
data["Marital_Status"] = data["Marital_Status"].astype(str).str.replace("0", "Single")
```

```python
data
```

- **Dropping columns(Status,unnamed1) as they have null values.**

```python
data = data.drop({"unnamed1","Status"}, axis=1)
```

- **Drop all null values from data.**

```python
data = data.dropna()
```

```python
data.head()
```

# *Countplot (Age group wise gender count)*

```python
colors = [ 
    "#F3E5F5", 
    "#E1BEE7", 
    "#CE93D8", 
    "#BA68C8", 
    "#AB47BC", 
    "#9C27B0", 
    "#8E24AA", 
    "#6A1B9A", 
    "#4A148C", 
] 
```

```python
plt.figure(figsize=(10,6))

c1 = ["#5C1E65","#9117DE"]

A1 = sns.countplot(x="Age Group",data=data,hue="Gender",palette=c1,edgecolor="black")

plt.title("Age group wise gender count",family = "georgia",color="#2C0144",fontsize=16,fontweight="bold")

plt.xlabel("")

plt.ylabel("")

for y in A1.containers:
    A1.bar_label(y)
```

***Explanation:***
The count plot defines:
- The age group of 0 to 17 has the lowest number of people.
- The age group of 26 to 35 has the highest number of people especially females(3269).

# ***Line Chart (Sector wise Total Orders)***

```python
T1 = data.groupby("Sector",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False)
```

```python

plt.figure(figsize=(15,5))
sns.lineplot(x="Sector",y="Orders",data=T1,ci=None,marker="o",color="red")
plt.title("Sector wise Total orders",family = "georgia",color="#CA1B1B",fontsize=16,fontweight="bold")
plt.grid()
plt.xticks(rotation=45)
plt.xlabel("")
plt.ylabel("")

plt.show()
```

***Explanation:***
- The line chart conveys that IT sector has highest number of orders and Agriculture sector has the lowest number of orders.

# ***Pie chart(Zone wise Amount)***

```python
C2 = colors = [ 
    "#E8F5E9", 
    "#A5D6A7", 
    "#81C784", 
    "#4CAF50",  
    "#1B5E20", 
] 
```

```python
T2 = data.groupby("Zone",as_index=False)["Amount"].sum()

plt.figure(figsize=(8,6))

plt.pie(T2["Amount"],labels=T2["Zone"],autopct="%1.f%%",startangle=180,colors=C2)

plt.title("Zone wise amount",family = "georgia",color="#192BA3",fontsize=16,fontweight="bold")

plt.show()
```

***Explanation:***
- The Central zone has highest amount whereas, the Eastern zone has the lowest amount.

```python
C3 = colors = [ 
    "#C5535E", 
    "#FFE0B2", 
    "#BEB349", 
    "#C8E6C9", 
    "#B2EBF2", 
    "#BBDEFB", 
    "#494257", 
    "#EB78A0", 
] 
```
